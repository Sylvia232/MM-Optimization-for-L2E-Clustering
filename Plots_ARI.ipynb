```python
import matplotlib.pyplot as plt

# Set global plotting style BEFORE creating any plots
plt.rcParams.update({
    "font.size": 14,
    "axes.titlesize": 16,
    "axes.labelsize": 14,
    "legend.fontsize": 12,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12
})
```

```python
import numpy as np

ari_mm = np.load("ari_mm_NSU14OCTdifwt.npy")#
ari_tkmeans = np.load("ari_tkmeans_NSU14OCTdifwt.npy")#excomp
ari_pam = np.load("ari_pam_NSU14OCT.npy")
ari_tmix = np.load("ari_tmix_NSU14OCTdifwt.npy")

# Find dataset columns where TMIX has -1 in any row
bad_datasets = np.any(ari_tmix == -1, axis=0)
#print(len(bad_datasets))
# Keep only good datasets
ari_mm_filt = ari_mm[:, ~bad_datasets]
ari_pam_filt = ari_pam[:, ~bad_datasets]
ari_tkmeans_filt = ari_tkmeans[:, ~bad_datasets]
ari_tmix_filt = ari_tmix[:, ~bad_datasets]

ari_mats= [
    ari_mm_filt,
    ari_pam_filt,
    ari_tkmeans_filt,
    ari_tmix_filt
]

#ari_mats = [ari_mm, ari_pam, ari_robustkmeanspp, ari_tmix]#, ari_tclust]
methods = ["MM", "PAM", "TKMEANS", "TMIX"]#, "TCLUST"]

means = []
stds = []

fractions = np.arange(0, 0.6, 0.1)
for mat in ari_mats:
    means.append(mat.mean(axis=1))  # shape (6,)
    stds.append(mat.std(axis=1))    # shape (6,)

plt.figure(figsize=(7,5))
for mean, std, method in zip(means, stds, methods):
    plt.plot(fractions, mean, marker='o', label=method)
    plt.fill_between(
        fractions,
        mean - std,
        mean + std,
        alpha=0.2
    )

plt.xlabel("Fraction of outliers")
plt.ylabel("ARI")
plt.title("Mean ARI with standard deviation")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("ARI_NSU14OCT.png", dpi=300, bbox_inches="tight")

plt.show()
```
